## 2.3 Manipulate string with string methods

```python
child_1 = "Jacob"
child_2 = "             Deborah               "
child_3 = "Pranciya    "
```

```python
len(child_1), len(child_2), len(child_3)
```

```python
print(child_1)
print(child_2)
print(child_3)
```

```python
child_1.isupper()
```

```python
child_3.count("a")
```

```python
child_1 = child_1.upper()
child_2 = child_2.strip()
child_3 = child_3.rstrip()
```

```python
print(child_1)
print(child_2)
print(child_3)
```

```python
len(child_1), len(child_2), len(child_3)
```

## 2.5 Chllenge-1

```python
input_colour = input("Enter your favourite colour")
input_number = input("Enter your favourite number")
print("Congratulation! you won "+input_number+" "+input_colour+" gifts!") # we used 4 + operation which will take lot of space. Time & Space Complexity - Data Structure Algorithm(DSA)
```

```python
input_colour = input("Enter your favourite colour")
input_number = input("Enter your favourite number")
print("Congratulation! you won ",input_number,input_colour," gifts!") #we used only comma, this is better solution for saving the space
```

## 2.6 Streamlining the print statement
Print all the information in one quote

```python
print(f"Congratulation! you won {input_number} {input_colour} gifts!") # { is placeholder
```

## 4.1 Write your own Function

### Three typs of Functions:
* **Predefined Functions/Built-in functions** - which will get automatically installed into your local machine when we install anaconda navigator.
* **User defined functions** - the function defined by user or customizable function
* **Anonymous function** - It is nameless function, which is used to accomplish one operation at once.

```python
def greeting():
    return "Jesus loves you"
```

```python
greeting()
```

### Parameters and Arguments

```python
def addition(num1,num2,num3): # The values inside brackets are parameters
    return num1+num2+num3
```

```python
addition(1,2,3) # the values inside brackets are arguments
```

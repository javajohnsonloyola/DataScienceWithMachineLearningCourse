# Level 2: Print if the user guessed number is correct or wrong with maximum of 7 attempts

## 2.1 Using For Loop - end iteration is defined

```python
secret_number = 56
for i in range(7):
    user_input = int(input("Enter your guess:"))
    if user_input == secret_number:
        print("Correct answer")
        break
    else:
        print("Wrong answer")
```

## 2.2 Using While Loop - when end number is not defined and the instruction has to be executed infinite times

```python
secret_number = 56
while True:
    user_input = int(input("Enter your guess:"))
    if user_input == secret_number:
        print("Correct answer")
        break
    else:
        print("Wrong answer")
```

# Level 3: Give multiple conditions

2. User should enter the number > secret number, then the program should say "Your number is greater than the secret number"
3. User should enter the number < secret number, then the program should say "Your number is less than the secret number"

```python
secret_number = 56
for i in range(7):
    user_input = int(input("Enter your guess:"))
    if user_input < 1 or user_input > 100:
        print("Your number must be between 1 and 100")
    elif user_input > secret_number:
        print("Your number is greater than the secret number")
    elif user_input < secret_number:
        print("Your number is less than the secret number")
    else:
        print("Correct answer")
        break
```

```python
secret_number = 56
for i in range(7):
    user_input = int(input("Enter your guess:"))
    if user_input < 1 or user_input > 100:
        print("Your number must be between 1 and 100")
    elif user_input > secret_number:
        print("Your number is greater than the secret number")
    elif user_input < secret_number:
        print("Your number is less than the secret number")
    else:
        break
if user_input == secret_number:
    print("Correct answer")
else:
    print(f"exceeded {i+1} attempts")
```

# Level 5: The Secret Number shouldn't be fixed, it should be random

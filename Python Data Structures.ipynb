## 1. List
* 1D
* Heterogeneous(collection of muliple data typs)
* Dupliate Values
* Indexing & Slicing is possible
* Mutable - Values can be changed

```python
empty_cart = []
```

```python
type(empty_cart)
```

```python
items_to_buy = ["Banana","Apple","Grape","Orange","Banana",8.9,43]
items_to_buy
```

```python
type(items_to_buy)
```

```python
len(items_to_buy)
```

```python
items_to_buy[3]
```

```python
items_to_buy[-5]
```

```python
items_to_buy[2:5] #last index is not included
```

```python
items_to_buy[-2:-1] #last index is not included
```

```python
items_to_buy[-2:] #to include last index, leave it blank
```

```python
items_to_buy[1] = "Watermelon" # changing value is possible, it is called mutable
```

```python
items_to_buy
```

```python
items_to_buy.append("Mango")
```

```python
items_to_buy
```

## 2. Sets

* 1D
* Heterogeneous(collection of muliple data typs)
* It won't allow duplicate values
* Indexing and Slicing is not possible(passible using functions)
* Mutable(using functions)

```python
empty_cart_2 = set()
empty_cart_2
```

```python
type(empty_cart_2)
```

```python
items_to_buy_2 = {"Banana","Apple","Grape","Orange","Banana",8.9,43}
items_to_buy_2
```

```python
type(items_to_buy_2)
```

```python
items_to_buy_2.add("Mango") # to add elements, need to use functions, after dot, press tab to see the available functions
```

```python
items_to_buy_2
```

```python
items_to_buy_2.discard("Banana")
```

```python
items_to_buy_2
```

## 3. Tuple
* 1D
* Heterogeneous(collection of muliple data typs)
* Duplicate values are allowed
* Indexing and Slicing is possible
* Immutable

```python
empty_cart_3 = ()
type(empty_cart_3)
```

```python
items_to_buy_3 = ("Apple","Orange","Grapes","Apple",6.76,8,9,"pomegranate")
items_to_buy_3
```

```python
items_to_buy_3[0:3]
```

```python
items_to_buy_3[0] = "Banana"
```

```python
items_to_buy_3.count("Apple")
```

```python
items_to_buy_3.index("Apple")
```

## 4. Dictionary
It is a Key:Value Pair
* 2D
* Heterogeneous(collection of muliple data typs)
* Mutable
* Indexing & Slicing are not possible
* Don't create duplicate keys, but you can create duplicate values

```python
empty_cart_4 = {}
empty_cart_4
```

```python
type(empty_cart_4)
```

```python
student_marks = {"maths":98,
                 "English": 90,
                 "Science": 88}
student_marks
```

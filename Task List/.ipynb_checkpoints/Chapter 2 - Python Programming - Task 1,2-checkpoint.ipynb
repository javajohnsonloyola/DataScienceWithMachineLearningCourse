# Task 1 - 1 Function Example for all Python Data Structures

## 1.1 List Functions

```python
sample_list = []
```

```python
# append()	Adds an element at the end of the list
sample_list = ["Johnson","Princy"]
sample_list
```

```python
sample_list.append("Jacob")
sample_list
```

```python
# clear()	Removes all the elements from the list
sample_list.clear()
sample_list
```

```python
# copy()	Returns a copy of the list
sample_list = ["Johnson","Princy"]
sample_list2 = sample_list.copy()
sample_list2
```

```python
# count()	Returns the number of elements with the specified value
sample_list = ["Johnson","Princy","Johnson"]
sample_list.count("Johnson")
```

```python
# extend()	Add the elements of a list (or any iterable), to the end of the current list
sample_list = ["Johnson","Princy"]
sample_list2 = ["Jacob"]
sample_list.extend(sample_list2)
sample_list
```

```python
sample_list.extend("Deborah") #If the value is added directly instead of a list, it is added one by one. So add as a list
sample_list
```

```python
# index()	Returns the index of the first element with the specified value
sample_list = ["Johnson","Princy","Jacob"]
sample_list.index("Jacob")
```

```python
# insert()	Adds an element at the specified position
sample_list = ["Johnson","Princy","Jacob"]
sample_list.insert(1,"Deborah")
sample_list
```

```python
#pop()	Removes the element at the specified position
sample_list = ["Johnson","Princy","Jacob"]
sample_list.pop(1) # argument is optional. default value is -1, which returns the last item
sample_list
```

```python
sample_list = ["Johnson","Princy","Jacob"]
sample_list.pop() # argument is optional. default value is -1, which returns the last item
sample_list
```

```python
# remove()	Removes the first item with the specified value
sample_list = ["Johnson","Princy","Jacob"]
sample_list.remove("Princy") #argument is required
sample_list
```

```python
# reverse()	Reverses the order of the list
sample_list = ["Johnson","Princy","Jacob"]
sample_list.reverse()
sample_list
```

```python
# sort()	Sorts the list
sample_list = ["Johnson","Princy","Jacob"]
sample_list.sort()
sample_list
```

## 1.2 Set Functions

```python
sample_set = set()
```

```python
# add() Adds an element to the set
sample_set = {"Johnson","Princy","Jacob"}
sample_set.add("Deborah")
sample_set
```

```python
# clear() Removes all the elements from the set
sample_set.clear()
sample_set
```

```python
# copy() Returns a copy of the set
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = sample_set.copy()
sample_set2
```

```python
# difference() Returns a set containing the difference between two or more sets
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set.difference(sample_set2) #Return a set that contains the items that only exist in set sample_set, and not in set sample_set2:
```

```python
# Use - as a shortcut instead of difference():
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set - sample_set2
```

```python
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set - sample_set2
sample_set
```

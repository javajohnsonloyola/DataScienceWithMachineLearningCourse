```python
text1 = "i am jacob"
text2 = "I Am jAcob"
text3 = "I am Jacob and I read lot of books, I and I spend lot of time in reading the books"
```

```python
#capitalize()	Converts the first character to upper case
text1 = "i am jacob"
print(text1.capitalize())
```

```python
#casefold()	Converts string into lower case
text2 = "I Am jAcob"
print(text2.casefold())
```

```python
# lower() Converts a string into lower case
text2 = "I Am jAcob"
print(text2.lower())
```

```python
#swapcase()	Swaps cases, lower case becomes upper case and vice versa
text2 = "I Am jAcob"
print(text2.swapcase())
```

```python
#title() Converts the first character of each word to upper case
text2 = "I Am jAcob"
print(text2.title())
```

```python
#upper() Converts a string into upper case
text2 = "I Am jAcob"
print(text2.upper())
```

```python
#count() Returns the number of times a specified value occurs in a string
text3 = "I am Jacob and I read lot of books, I and I spend lot of time in reading the books"
print(text3.count("books"))
```

```python
#find()	Searches the string for a specified value and returns the position of where it was found
text3 = "I am Jacob and I read lot of books, I and I spend lot of time in reading the books"
print(text3.find("a"))
print(text3.find("a",3,10)) #Where in the text is the first occurrence of the letter "a" when you only search between position 3 and 10
print(text3.find("books")) 
print(text3.find("x")) #If not found, returns -1
```

```python
#index()	Searches the string for a specified value and returns the position of where it was found
#index() method is almost the same as the find() method, the only difference is that the find() method returns -1 if the value is not found
text3 = "I am Jacob and I read lot of books, I and I spend lot of time in reading the books"
print(text3.index("a"))
print(text3.index("a",3,10)) #Where in the text is the first occurrence of the letter "a" when you only search between position 3 and 10
print(text3.index("books")) #It searches all letters, not the whole text
print(text3.index("x")) #If not found, returns ValueError
```

```python
#rfind()	Searches the string for a specified value and returns the last position of where it was found
text3 = "I am Jacob and I read lot of books, I and I spend lot of time in reading the books"
print(text3.rfind("a"))
print(text3.rfind("a",3,10)) #Where in the text is the first occurrence of the letter "a" when you only search between position 3 and 10
print(text3.rfind("books"))
print(text3.rfind("x")) #If not found, returns -1
```

```python
#rindex()	Searches the string for a specified value and returns the last position of where it was found
#rindex() method is almost the same as the rfind() method, the only difference is that the rfind() method returns -1 if the value is not found
text3 = "I am Jacob and I read lot of books, I and I spend lot of time in reading the books"
print(text3.rindex("a"))
print(text3.rindex("a",3,10)) #Where in the text is the first occurrence of the letter "a" when you only search between position 3 and 10
print(text3.rindex("books")) #It searches all letters, not the whole text
print(text3.rindex("x")) #If not found, returns ValueError
```

```python
#3.1 isalnum()	Returns True if all characters in the string are alphanumeric
text4 = "Jacobis10"
text5 = "Jacob is 10"
print(text4.isalnum())
print(text5.isalnum())
```

```python
#3.2 isalpha()	Returns True if all characters in the string are in the alphabet
text4 = "Jacob"
text5 = "Jacobis10"
print(text4.isalpha())
print(text5.isalpha())
```

```python
#3.3 isascii()	Returns True if all characters in the string are ascii characters (a-z)
text4 = "Jacob10&%?[]"
print(text4.isascii())
```

```python
#3.4 isdecimal()	Returns True if all characters in the string are decimals
text1 = "1.8" # dot is not decimal, so returns False
text2 = "1"
print(text1.isdecimal())
print(text2.isdecimal())
```

```python
#3.5 isdigit()	Returns True if all characters in the string are digits
text1 = "1.8" # dot is not digit, so returns False
text2 = "1"
print(text1.isdigit())
print(text2.isdigit())
```

```python
#3.6 isidentifier()	Returns True if the string is an identifier
#A string is considered a valid identifier if it only contains alphanumeric letters (a-z) and (0-9), or underscores (_). A valid identifier cannot start with a number, or contain any spaces.
text1 = "MySon"
text2 = "MySonAgeIs10"
text3 = "10Age"
text4 = "My Son"

print(text1.isidentifier())
print(text2.isidentifier())
print(text3.isidentifier())
print(text4.isidentifier())
```

```python
#3.7 islower()	Returns True if all characters in the string are lower case
text1 = "my son name is jacob"
text2 = "My son name is jacob"
print(text1.islower())
print(text2.islower())
```

```python
#3.8 isnumeric()	Returns True if all characters in the string are numeric
#"-1" and "1.5" are NOT considered numeric values, because all the characters in the string must be numeric, and the - and the . are not.
text1 = "56"
text2 = "5.6"
print(text1.isnumeric())
print(text2.isnumeric())
```

```python
#3.9 isprintable()	Returns True if all characters in the string are printable
text1 = "Jacob is 10"
text2 = "Jacob \n is 10"
print(text1.isprintable())
print(text2.isprintable())
```

```python
#3.10 isspace()	Returns True if all characters in the string are whitespaces
text1 = "   "
text2 = " Jacob "
print(text1.isspace())
print(text2.isspace())
```

```python
#3.11 The istitle() method returns True if all words in a text start with a upper case letter, AND the rest of the word are lower case letters, otherwise False.
text1 = "HELLO, AND WELCOME TO MY WORLD"
text2 = "Hello"
text3 = "22 Names"
text4 = "This Is %'!?"
print(text1.istitle())
print(text2.istitle())
print(text3.istitle())
print(text4.istitle())
```

```python
#3.12 isupper()	Returns True if all characters in the string are upper case
text1 = "Hello Jacob!"
text2 = "hello 123"
text3 = "MY NAME IS JACOB"

print(text1.isupper())
print(text2.isupper())
print(text3.isupper())
```

```python
# strip()	Returns a trimmed version of the string
text1 = "   Jacob   " #removes any leading, and trailing whitespaces
text2 = ",,,,,cttgg.....Jacob....rrr" #Remove the leading and trailing characters
print(text1.strip())
print(text2.strip(",.crtg"))
```

```python
#lstrip()	Returns a left trim version of the string
text1 = "   Jacob "
text2 = ",,,,rctt Jacob rctt"
print(text1.lstrip())
print(text2.lstrip(" ,.ctr"))
```

```python
#replace()	Returns a string where a specified value is replaced with a specified value
text = "Jacob is the son of Johnson. Jacob is 10 years old, Jacob is a good boy."
print(text.replace("Jacob","Johnson"))
print(text.replace("Jacob","Johnson",2)) #Replace the two first occurrence of the word "Jacob"
```

```python
# translate()	Returns a translated string
# maketrans()	Returns a translation table to be used in translations
text = "Hello, Son!"
mytable = str.maketrans("Son", "Jac") #Number of charactor should be equal, here its 3
print(text.translate(mytable))
```

```python
#split()	Splits the string at the specified separator, and returns a list
text1 = "My son name is Jacob, his age is 10, And he loves to read books"
print(text1.split())
print(text1.split(","))
print(text1.split(",",1)) # setting the maxsplit parameter to 1, will return a list with 2 elements!
```

```python
# rsplit()	Splits the string at the specified separator, and returns a list
text1 = "My son name is Jacob, his age is 10, And he loves to read books"
print(text1.rsplit(",",1)) # setting the maxsplit parameter to 1, will return a list with 2 elements(from right)
```

```python
# splitlines()	Splits the string at line breaks and returns a list
text1 = "My son name is Jacob\n his age is 10\n And he loves to read books"
print(text1.splitlines())
print(text1.splitlines(True)) #Specifies if the line breaks should be included (True), or not (False)
print(text1.splitlines(False))
```

```python
#partition()	Returns a tuple where the string is parted into three parts
text1 = "My son name has 2 apples and 3 bananas and 5 grapes"
print(text1.partition("and")) 
#1 - everything before the "match"
#2 - the "match"
#3 - everything after the "match"
print(text1.partition("or")) #If the specified value is not found, the partition() method returns a tuple containing: 1 - the whole string, 2 - an empty string, 3 - an empty string:
```

```python
# rpartition()	Returns a tuple where the string is parted into three parts
text1 = "My son name has 2 apples and 3 bananas and 5 grapes"
print(text1.rpartition("and")) 
print(text1.rpartition("or")) #If the specified value is not found, the partition() method returns a tuple containing: 1 - the whole string, 2 - an empty string, 3 - an empty string:
```

```python
# join()	Converts the elements of an iterable into a string
myTuple  = ("apple","grapes","orange")
print("#".join(myTuple))

myDict = {"name":"Jacob","age":"10","place":"Johannesburg"}
mySeperator = "#"
print(mySeperator.join(myDict))
```

```python
# center()	Returns a centered string
text1 = "jacob"
print(text1.center(20))
```

```python
#ljust()	Returns a left justified version of the string
#rjust()	Returns a right justified version of the string
text1 = "jacob"
print(text1.ljust(20),"is my son's name")
print(text1.rjust(20),"is my son's name")
```

```python
#zfill()	Fills the string with a specified number of 0 values at the beginning
text1 = "jacob"
text2 = "jacob is"
print(text1.zfill(10))
print(text2.zfill(10))
```

```python
# format()	Formats specified values in a string
print("My name is {}, and I'm {}".format("Johnson","38"))
```

```python
# format_map()	Formats specified values from a dictionary in a string
my_var = {"name":"Johnson","age":"38"}
print("My name is {name}, and I'm {age}".format_map(my_var))
```

```python
# f-strings To specify a string as an f-string, simply put an f in front of the string literal, and add curly brackets {} as placeholders for variables and other operations.
name = "Johnson"
age = 38
print(f"My name is {name}, and I'm {age}")
```

```python
#len()	Returns the length of an object
text = "Jacob"
print(len(text))
```

```python
# str()	Returns a string object
no_of_children = 2
print("I have "+str(no_of_children)+" children")
print("I have "+no_of_children+" children") #throw an error because of the type
```

```python
# 13.3 slice() function is used to create a slice object that specifies how to extract a portion of a sequence such as a string, list or tuple. 
#slice(start, stop, step)
text = "My name is Jacob"
print(text[slice(0,10)],"Johnson")
my_list = ['a','b','c','d','e']
print(my_list[slice(0,3)])
my_tuple = ("a","b","c","d","e","f","g")
print(my_tuple[slice(0,8)])
print(my_tuple[slice(0,8,2)]) #step is 2
print(my_tuple[slice(-1,-8,-1)]) #backwards
```

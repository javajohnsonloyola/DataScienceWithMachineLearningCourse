```python
# 1. while loop - example 1
total_pages = 5
page = 1
while True:
    if total_pages < page:
        break
    print(page)
    page+=1
```

```python
# 1. while loop - example 2
my_list = ["Johnson","Princy","Jacob","Deborah","End"]
count = 0
while True:
    print(my_list[count])
    # print(my_list[count].find("End"))
    if my_list[count] == "End":
        break
    count+=1
```

```python
# 2. for loop - example 1
total_pages = 5
for i in range(1,total_pages+1):
    print(i)
```

```python
# 2. for loop - example 2
my_list = ["Johnson","Princy","Jacob","Deborah","End"]
for i in range(len(my_list)):
    print(my_list[i])
```

```python
# 3. if else - example 1
total_pages = 5
for i in range(1,total_pages+1):
    if(i == 5):
        print("page",i,"is the end")
    else:
        print("page",i,"is not the end")
```

```python
# 3. if else - example 2
my_list = ["Johnson","Princy","Jacob","Deborah","End"]
for i in range(len(my_list)):
    if(my_list[i] == "Johnson"):
        print("Hello, Dad!")
    else:
        print("Hello, Johnson's family!")
```

```python
# 4. if elif else - example 1
total_pages = 5
for i in range(1,total_pages+1):
    if(i == 5):
        print("page",i,"is the end")
    elif(i == 1):
        print("you are on page",i,"and you just started")
    else:
        print("page",i,"is not the end")
```

```python
# 4. if elif else - example 2
my_list = ["Johnson","Princy","Jacob","Deborah","End"]
for i in range(len(my_list)):
    if(my_list[i] == "Johnson"):
        print("Hello, Dad!")
    elif(my_list[i] == "Princy"):
        print("Hello, Mummy!")
    else:
        print("Hello, Johnson's family!")
```

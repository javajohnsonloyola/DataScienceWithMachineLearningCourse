Rock, Paper, Scissors Game
Objective:
Build a simple Rock-Paper-Scissors game where the player competes with the computer.
 This task will test your understanding of:
●	Input/output
●	Conditional logic
●	Random module
●	Loops
●	Exception handling

Instructions:
1.	The user should enter their choice: rock, paper, or scissors.
2.	The computer should randomly select one of these.
3.	Compare both and print who wins.
4.	Ask if the player wants to play again- if yes, repeat; if no, exit.
5.	Handle invalid inputs gracefully using try-except or condition checks.


```python
import random
mylist = ["rock","paper","scissors"]
while True:
    computer_choice = random.choice(mylist)
    try:
        user_choice = input("Enter your choice, your choices are rock, paper, scissors: ")
        if user_choice not in mylist:
            raise ValueError("invalid choice")
    except ValueError:
        print("Invalid choice")
        continue
    if(computer_choice == "rock" and user_choice == "rock"):
        print("computer choice is",computer_choice," you tied")
    elif(computer_choice == "rock" and user_choice == "paper"):
        print("computer choice is",computer_choice,", you won!")
    elif(computer_choice == "rock" and user_choice == "scissors"):
        print("computer choice is",computer_choice,", you lost!")
    
    elif(computer_choice == "paper" and user_choice == "rock"):
        print("computer choice is",computer_choice,", you lost!")
    elif(computer_choice == "paper" and user_choice == "paper"):
        print("computer choice is",computer_choice,", you tied")
    elif(computer_choice == "paper" and user_choice == "scissors"):
        print("computer choice is",computer_choice,", you won!")
        
    elif(computer_choice == "scissors" and user_choice == "rock"):
        print("computer choice is",computer_choice,", you won!")
    elif(computer_choice == "scissors" and user_choice == "paper"):
        print("computer choice is",computer_choice,", you lost!")
    elif(computer_choice == "scissors" and user_choice == "scissors"):
        print("computer choice is",computer_choice,", you tied")
    else:
        print("computer choice is ",computer_choice,", I'm confused")
    try:
        is_repeat = input("Do you want to play again? yes/no:")
        if is_repeat not in ["yes","no"]:
            raise ValueError("invalid respose")
    except ValueError:
        print("Invalid respose")
    if is_repeat == "no":
        print("Thank you!")
        break
    
```

# Task 4
Word Guessing Game (Mini Hangman)
Objective:
Create a word-guessing game where the player tries to guess the letters of a hidden word.

This will test your:
●	Loops
●	Conditional logic
●	String manipulation
●	Exception handling
●	Logical thinking

Instructions:
1.	Create a list of random words (like “python”, “machine”, “learning”, “bridge”).
2.	Randomly select one word for the game.
3.	Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
4.	Keep track of wrong guesses (max 6 attempts).
5.	End the game when the player either guesses the full word or uses all chances.


```python
import random
mylistword = ["jacob","johnson","pranciya","deborah"]
myword = random.choice(mylistword)
original_word = myword
wrong_guess = 0
correctly_guessed_letters = ""
while True:
    display = ""
    for letter in original_word:
        if letter in correctly_guessed_letters:
            display += letter + " "
        else:
            display += "_ "
    print(display)
    try:
        guessed_letter = input("guess the letter")
        if len(guessed_letter) != 1:
            raise ValueError("Enter only one letter")
    except ValueError as e:
        print(e)
        continue
    if guessed_letter in myword:
        myword = myword.replace(guessed_letter, "")
        print("correct")
        correctly_guessed_letters += guessed_letter
    else:
        print("wrong")
        wrong_guess +=1
    if myword == "" or wrong_guess >= 6:
        break
if myword == "":
    print("you guessed the full word")    
else:
    print("you used all chances")
    
```

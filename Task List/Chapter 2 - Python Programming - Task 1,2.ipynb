## 1.2 Set Functions
https://www.w3schools.com/python/python_ref_set.asp

```python
sample_set = set()
```

```python
# add() Adds an element to the set
sample_set = {"Johnson","Princy","Jacob"}
sample_set.add("Deborah")
sample_set
```

```python
# clear() Removes all the elements from the set
sample_set.clear()
sample_set
```

```python
# copy() Returns a copy of the set
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = sample_set.copy()
sample_set2
```

```python
# difference() Returns a set containing the difference between two or more sets (Shortcut -)
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set.difference(sample_set2) #Return a set that contains the items that only exist in set sample_set, and not in set sample_set2:
```

```python
# Use - as a shortcut instead of difference():
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set3 = sample_set - sample_set2
sample_set3
```

```python
# difference_update() Removes the items in this set that are also included in another, specified set (Shortcut -=)
# The difference() method returns a new set, without the unwanted items, and the difference_update() method removes the unwanted items from the original set.
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set.difference_update(sample_set2) #the removes "Johnson" from the original set. Not return the difference like difference() method
sample_set
```

```python
# Use -= as a shortcut instead of difference_update():
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set -= sample_set2
sample_set
```

```python
# discard()	 Remove the specified item
sample_set = {"Johnson","Princy","Jacob"}
sample_set.discard("Princy")
sample_set
```

```python
# intersection() Returns a set, that is the intersection of two other sets (Shortcut &)
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set3 = sample_set.intersection(sample_set2)
sample_set3
```

```python
# Use & as a shortcut instead of intersection():
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set3 = sample_set & sample_set2
sample_set3
```

```python
# intersection_update() Removes the items in this set that are not present in other, specified set(s) (Shortcut &=)
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set.intersection_update(sample_set2)
sample_set
```

```python
# Use &= as a shortcut instead of intersection_update():
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set &= sample_set2
sample_set
```

```python
# isdisjoint() Returns whether two sets have a intersection or not
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set3 = sample_set.isdisjoint(sample_set2) #Returns False because "Johnson" is present in sample_set2
sample_set3
```

```python
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Deborah"}
sample_set3 = sample_set.isdisjoint(sample_set2) #Returns True because no items in sample_set is present in sample_set2
sample_set3
```

```python
# issubset() Returns True if all items of this set is present in another set (Shortcut <=)
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Princy"}
sample_set3 = sample_set.issubset(sample_set2)
sample_set3
```

```python
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Princy"}
sample_set3 = sample_set2.issubset(sample_set)
sample_set3
```

```python
# Shortcut <=
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Princy"}
sample_set3 = sample_set2 <= sample_set
sample_set3
```

```python
# Shortcut <
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Princy"}
sample_set3 = sample_set2 < sample_set
sample_set3
```

```python
# issuperset() Returns True if all items of another set is present in this set (Shortcut >=)
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Princy"}
sample_set3 = sample_set.issuperset(sample_set2)
sample_set3
```

```python
# Shortcut >=
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Princy"}
sample_set3 = sample_set >= sample_set2
sample_set3
```

```python
# Shortcut >
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Princy"}
sample_set3 = sample_set > sample_set2
sample_set3
```

```python
# pop() removes a random item from the set
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = sample_set.pop()
sample_set
```

```python
# remove() Removes the specified element
sample_set = {"Johnson","Princy","Jacob"}
sample_set.remove("Johnson")
sample_set
```

```python
# symmetric_difference() Returns a set with the symmetric differences of two sets (Shortcut ^)
# Return a set that contains all items from both sets, except items that are present in both sets:
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set3 = sample_set.symmetric_difference(sample_set2)
sample_set3
```

```python
# Shortcut ^
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set3 = sample_set ^ sample_set2
sample_set3
```

```python
# symmetric_difference_update()	Inserts the symmetric differences from this set and another (Shortcut ^=)
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set.symmetric_difference_update(sample_set2)
sample_set
```

```python
# Shortcut ^=
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set ^= sample_set2
sample_set
```

```python
# union() Return a set containing the union of sets (Shortcut | )
#Return a set that contains all items from both sets, duplicates are excluded
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set3 = sample_set.union(sample_set2)
sample_set3
```

```python
# Shortcut |
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set3 = sample_set | sample_set2
sample_set3
```

```python
# update() Update the set with the union of this set and others (Shortcut |=)
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set.update(sample_set2)
sample_set
```

```python
# Shortcut |=
sample_set = {"Johnson","Princy","Jacob"}
sample_set2 = {"Johnson","Deborah"}
sample_set |= sample_set2
sample_set
```

## 1.4 Dictionary Functions

```python
sample_dict = {}
type(sample_dict)
```

```python
# clear() Removes all the elements from the dictionary
sample_dict = {"name":"Jacob","age":10,"place":"Johannesburg"}
sample_dict
```

```python
sample_dict.clear()
sample_dict
```

```python
# copy() Returns a copy of the dictionary
sample_dict = {"name":"Jacob","age":10,"place":"Johannesburg"}
sample_dict2 = sample_dict.copy()
sample_dict2
```

```python
# fromkeys() Returns a dictionary with the specified keys and value
# Create a dictionary with 3 keys, all with the value 0:
sample_keys = ("name","age","place")
init_values = 0
sample_dict = dict.fromkeys(sample_keys,init_values)
sample_dict
```

```python
# get()	Returns the value of the specified key
sample_dict = {"name":"Jacob","age":10,"place":"Johannesburg"}
sample_dict.get("place")
```

```python
# items() Returns a list containing a tuple for each key value pair
sample_dict = {"name":"Jacob","age":10,"place":"Johannesburg"}
sample_dict.items()
```

```python
# keys() Returns a list containing the dictionary's keys
sample_dict = {"name":"Jacob","age":10,"place":"Johannesburg"}
sample_dict.keys()
```

```python
# pop() Removes the element with the specified key
sample_dict = {"name":"Jacob","age":10,"place":"Johannesburg"}
sample_dict.pop("age")
sample_dict
```

```python
# popitem() Removes the last inserted key-value pair
#The popitem() method removes the item that was last inserted into the dictionary. In versions before 3.7, the popitem() method removes a random item.
sample_dict = {"name":"Jacob","age":10,"place":"Johannesburg"}
sample_dict.popitem()
sample_dict
```

```python
# setdefault() Returns the value of the specified key. If the key does not exist: insert the key, with the specified value
sample_dict = {"name":"Jacob","age":10,"place":"Johannesburg"}
sample_dict2 = sample_dict.setdefault("school","Christ Preparatory Church School")
sample_dict2
```

```python
# update()	Updates the dictionary with the specified key-value pairs
sample_dict = {"name":"Jacob","age":10,"place":"Johannesburg"}
sample_dict.update({"School":"Christ Church Preparatory School"})
sample_dict
```

```python
# values()	Returns a list of all the values in the dictionary
sample_dict = {"name":"Jacob","age":10,"place":"Johannesburg"}
sample_dict.values()
```
